Ques-4 Write a program that asks the user to input a number and prints whether the number is positive and even, positive and odd, or negative.

```python
num = int(input("Enter a number: "))
if num > 0:
    if num % 2 == 0:
        print("Positive and Even")
    else:
        print("Positive and Odd")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

Ques-8 Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.

```python
total = 0
while True:
    num = int(input("Enter a number: "))
    if num == 0:
        break
    total += num
print("Sum =", total)
```

Ques-12 Write a program that calculates the factorial of a number input by the user using a while loop.

```python
num = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= num:
    factorial = factorial * i
    i += 1
print("Factorial =", factorial)
```

Ques-13 Write a program that calculates the sum of the digits of a number input by the user using a while loop.

```python
num = int(input("Enter a number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10
print("Sum of digits =", sum)
```

Ques-14 Write a program that checks if a number input by the user is a prime number using a for loop.

```python
num = int(input("Enter a number: "))
if num < 2:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```

Ques-15 Write a program that prints the first n Fibonacci numbers, where n is input by the user.

```python
n = int(input("Enter n: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```
